**3.Standarized text values**

```python
df['Education'] = df['Education'].str.strip().str.title()
df['Marital_Status'] = df['Marital_Status'].str.strip().str.title()

# Check the values
print("Education:", df['Education'].unique())
print("Marital_status:", df['Marital_Status'].unique())
```

```python
# want to standardize inconsistent category names
df['Marital_Status'] = df['Marital_Status'].replace({
    'Alone': 'Single',
    'Absurd': 'Other',
    'YOLO': 'Other'
})

print(df['Marital_Status'].unique())
```

```python
# Apply standardization to all text columns
# Find all text columns
text_columns = df.select_dtypes(include='object').columns

# Remove extra spaces and standardize capitalization
for column in text_columns:
    df[column] = df[column].str.strip().str.title()

print(df[text_columns].head())
```

**4. Rename column headers to clean and consisent format**

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)
df.columns
```
